```python
import pandas as pd
df = pd.read_csv(
"https://raw.githubusercontent.com/papayuh/datn1370-datasets/main/data/synthetic/coffee_shop_sales.csv")
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.regplot(data=df, x="temp_high_f", y="hot_drinks")
plt.title("How Does Hot Weather Affect Hot Drink Sales")
plt.xlabel("Daily High Temperature")
plt.ylabel("Number of Hot Drinks Sold")
plt.show()
```

Hot weather does affect hot drink sales. As the temperature increases the number of sales decreases. I used a scatterplot because I wanted to see the sales, but I also wanted to see the corresponding trendline to see pattern.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset so 'df' is defined in this session
df = pd.read_csv("https://raw.githubusercontent.com/papayuh/datn1370-datasets/main/data/synthetic/coffee_shop_sales.csv")

sns.boxplot(data=df, x="shop", y="revenue_usd")
plt.title("Revenue by Location")
plt.xlabel("Location")
plt.ylabel("Revenue in USD")
plt.show()
```

The Downtown location has the highest median daily revenue of approximately $1250, and the Campus location has the lowest median daily revenue at approximately $900. Riverside median revenue is around $1,000, but it does sometimes have random days that have unusual sales days of over $1,600. I chose a box plot because I wanted to be able to compare the daily revenue at the different locations, looking at its usual daily sales patterns.
